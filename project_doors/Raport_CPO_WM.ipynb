## Wymagania techniczne
Każdy projekt musi wykorzystywać i **uzasadniać** zastosowanie poniższych elementów przetwarzania obrazu:

1. **Przestrzenie barw:**  
    konwersja pomiędzy BGR ↔ HSV / YCbCr / Lab  
    należy użyć co najmniej jednej przestrzeni poza BGR
    *uzasadnij wybór – dlaczego akurat ta przestrzeń jest korzystna dla Twojego problemu*

2. **Operacje arytmetyczne / normalizacja:**  
    *pokaż wpływ na histogram lub wizualnie*

3. **Operacje geometryczne:**  
    np. korekcja perspektywy lub skalowanie / rotacja (z uzasadnieniem)  
    *można zilustrować na kilku przykładach*

4. **Filtracja:**  
    *wyjaśnij, dlaczego dobrałeś taki filtr i taki rozmiar kernela*

5. **Morfologia:**  
    zastosuj co najmniej dwie operacje (np. otwarcie + zamknięcie, dylatacja, erozja)  
    *pokaż wpływ na segmentację lub czyszczenie tła*

6. **Segmentacja:**  
    zaprezentuj metodę segmentacji (progowanie, analiza komponentów, klasteryzacja itp.)  
    *uzasadnij progi i parametry*

7. **Rozpoznawanie / wnioskowanie:**  
    np. dopasowanie szablonu, analiza komponentów, prosta klasyfikacja, metryki kształtu  
    *zademonstruj poprawne i błędne przypadki*

8. **Ewaluacja ilościowa:**  
    oceń skuteczność metod na zbiorze testowym  
    *docelowo ≥ 95 % skuteczności (lub metryka odpowiednia dla danego zadania)*  
    dodaj wykresy, tabelę wyników i krótką analizę błędów

```python
#Mój końcowy kod do trenowania

import os
from ultralytics import YOLO
# import torch # Opcjonalnie, jeśli chcesz sprawdzić, czy CUDA działa

if __name__ == '__main__':
    
    # 1. Definicja stałych i ścieżek
    
    # Bezwzględna ścieżka do Twojego głównego katalogu projektu
    PROJECT_DIR = 'F:/NoweProgramowanie/Studia/CPO/project_doors'
    
    # Pełna ścieżka do pliku konfiguracyjnego dataset.yaml
    DATA_YAML_PATH = os.path.join(PROJECT_DIR, 'dataset.yaml')
    
    # Sprawdzenie, czy plik istnieje
    if not os.path.exists(DATA_YAML_PATH):
        print(f"Błąd: Nie znaleziono pliku konfiguracyjnego w ścieżce: {DATA_YAML_PATH}")
        print("Upewnij się, że plik dataset.yaml został utworzony!")
        exit()
    
    # 2. Załadowanie modelu
    
    # ZMIENIONO: Ładowanie modelu YOLOv8s (small)
    model = YOLO('yolov8s.pt') 
    
    print(f"Rozpoczynanie treningu dla zbioru: {DATA_YAML_PATH}")
    # ZMIENIONO: Komunikat modelu bazowego
    print("Model bazowy: YOLOv8s (small)")
    
    # 3. Rozpoczęcie treningu
    
    results = model.train (
        data=DATA_YAML_PATH,  # Ścieżka do Twojego pliku konfiguracyjnego
        epochs=500,           # Liczba epok (możesz zmienić)
        imgsz=1024,            # Rozmiar wejściowy obrazu
        # UWAGA: Przy YOLOv8s i imgsz=1024, BATCH=24 może być za duży!
        # Rozważ zmniejszenie batcha do 16 lub 12, jeśli wystąpi błąd VRAM.
        batch=16,             
        name='drzwi_projekt_cpov8_s', # Dodaj '_s' do nazwy dla porządku
        workers=0,            # Zmienione na 0 dla Windows
        device=0,
        # 1. OPTYMALIZATOR
        optimizer='AdamW',         # Lepsza stabilność i konwergencja
        lr0=0.001,                 # Mniejsza szybkość uczenia dla precyzji
        # 2. FUNKCJA STRATY
        cls=1.5,                   # Zwiększona waga błędu klasyfikacji (na błąd mylenia klas)
        # 3. WZMOCNIENIE DANYCH
        hsv_v=0.15,                # Zwiększona zmiana jasności/oświetlenia
        translate=0.2,             # Większe losowe przesunięcia
        degrees=5.0                # Lekkie rotacje            
    )
    
    print("\n--- Trening zakończony ---")
```

### 6.3. Refleksja „co zadziałało / co nie”
Krótko podsumuj, które sugestie AI były pomocne, a które błędne lub nieprzydatne.

Na początku AI bardzo mi pomogło z samym ustawieniem i konfiguracją. Miałem dużo błędów związaych z bibliotekami i np. podpowiedzial, żebym dodał to "if __name__ == '__main__':" jeśli używam windows. Moje próby zaimplementowania korekcji obrotu z AI były nie skuteczne, więc ostatecznie z niego zrezygnowałem i wtedy postawnowiłem przejść na lepszy model sieci i użyć dodatkowych parametrów do trenowania. Ogólnie to obracanie raz działało raz nie, dlatego ostatacznie go nie ma. Totalnie też nie wiedziałem jak się trenuje sieć więc tutaj AI było bardzo pomocne, bo bez tego było by ciężko. AI też pomogło mi zapisywać do pliku najważniejsze informacje oraz obrazy w trakcie i po deketcji.
